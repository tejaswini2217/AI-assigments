```python
import pandas as pd
import heapq

# Load dataset
df = pd.read_csv("indian-cities-dataset.csv")

# Create graph
graph = {}

for _, row in df.iterrows():
    a = row["Origin"]
    b = row["Destination"]
    distance = row["Distance"]

    if a not in graph:
        graph[a] = []

    if b not in graph:
        graph[b] = []

    graph[a].append((b, distance))
    graph[b].append((a, distance))


# Dijkstra Algorithm
def dijkstra(start, goal):

    distances = {city: float("inf") for city in graph}
    previous = {city: None for city in graph}

    distances[start] = 0

    queue = [(0, start)]

    while queue:

        current_distance, current = heapq.heappop(queue)

        if current == goal:
            break

        for neighbour, distance in graph[current]:

            new_distance = current_distance + distance

            if new_distance < distances[neighbour]:

                distances[neighbour] = new_distance
                previous[neighbour] = current

                heapq.heappush(queue, (new_distance, neighbour))

    # Create path
    path = []
    current = goal

    while current is not None:
        path.append(current)
        current = previous[current]

    path.reverse()

    return path, distances[goal]


# Take input
start = input("Enter start city: ").strip()
goal = input("Enter goal city: ").strip()

# Correct common spelling
if start.lower() == "visakhapatnam":
    start = "Vishakhapatnam"

if goal.lower() == "visakhapatnam":
    goal = "Vishakhapatnam"

# Check cities
if start not in graph or goal not in graph:
    print("City not found in dataset")

else:
    path, distance = dijkstra(start, goal)

    print("\nShortest Path:")
    print(" -> ".join(path))

    print("Total Distance:", distance, "km")
```
